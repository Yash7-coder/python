```python
a=int(input("Enter the quantity of item1 : "))
b=int(input("Enter the quantity of item2 : "))
c=int(input("Enter the quantity of item3 : "))
p=int(input("Enter cost of item1 : "))
q=int(input("Enter cost of item1 : "))
r=int(input("Enter cost of item1 : "))
m=a*p
y=b*q
z=c*r
t=m+y+z
if t>50:
  x=t-0.1*t
  print("The Total Discounted Cost is : ",x)
else:
    print("The Total Cost is : ",t)
```
